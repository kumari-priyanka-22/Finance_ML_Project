FEATURING CRETING THAT WE HAVE DONE TILL NOW, MANY OF THEM HAS DATA LEAKAGE ISSUE(USING TARGET DATA/ FUTURE DATA) TO CREATE THE FEATURE
SO WE TAILOR THE FEATURES

```python
# PREVIOUS TRANSACTION AMOUNT
# Fraud often creates sudden spending jumps.
    
model_df = model_df.sort_values(
    ['client_id', 'date']
)
model_df['previous_amount'] = (
    model_df
    .groupby('client_id')['amount']
    .shift(1)
)
```

```python
# Compare current transaction against immediately previous one.
model_df['amount_change_ratio'] = (
    model_df['amount']
    /
    (model_df['previous_amount'] + 1)   # =1 TO AVOID DIVISION BY 0
)
```

```python
# Time Since Last Transaction
model_df['hours_since_last_txn'] = (
    model_df.groupby('client_id')['date']
    .diff()
    .dt.total_seconds()
    / 3600
)
```

```python
# Fraudsters frequently switch merchants.
model_df['previous_merchant'] = (
    model_df
    .groupby('client_id')['merchant_id']
    .shift(1)
)
```

```python
model_df['merchant_changed'] = (
    model_df['merchant_id']
    != model_df['previous_merchant']
).astype(int)
```

```python
model_df[
    [
        'previous_amount',
        'amount_change_ratio',
        'hours_since_last_txn',
        'merchant_changed'
    ]
].describe()
```

```python
# amount_changed_ratio is highly skewed (mean = 2.95, meadian = 0.8, max = 831) , outliers in data
model_df['amount_change_ratio_capped'] = (
    model_df['amount_change_ratio']
    .clip(upper=20)
)
model_df['amount_change_ratio_capped'].mean()
```

```python
# values are high, so transforming in log
model_df['log_hours_since_last_txn'] = np.log1p(
    model_df['hours_since_last_txn']
)
model_df['log_hours_since_last_txn'].describe()
```

```python
model_df['amount_change'] = (
    model_df['amount']
    - model_df['previous_amount']
)
model_df['amount_change'].describe()
```

```python
merchant_fraud = (
    model_df
    .groupby('merchant_changed')['fraud']
    .agg(['count','sum','mean'])
)

print(merchant_fraud)
```

```python
#can't do random split
# we use past data to predict future data target
# 2010-2017  → Train
# 2018       → Validation(why), cuz You tune the model repeatedly using the test set. Eventually the test set becomes part of training.
# 2019       → Test
```
