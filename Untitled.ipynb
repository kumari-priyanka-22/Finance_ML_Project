checking Outliers

```python
Q1 = transactions['amount'].quantile(0.25)
Q3 = transactions['amount'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

lower, upper
```

```python
(
    (transactions['amount'] < lower) |
    (transactions['amount'] > upper)
).mean() * 100
```
